```python
# outliers
def out_percent(arr):
    arr = arr[np.isfinite(arr)]
    
    q1 = np.percentile(arr,25)
    q3 = np.percentile(arr,75)
    iqr  = q3-q1
    lb = q1-1.5*iqr
    ub = q3+1.5*iqr
    
    outliers = ((arr > ub) | (arr < lb)).mean() 
    outliers_p = outliers*100
    return outliers_p
```

```python
credit_train_df[numerical_features_2].apply(out_percent).reset_index(name ="percentage_of outliers").sort_values(by="percentage_of outliers",ascending =False)
```
